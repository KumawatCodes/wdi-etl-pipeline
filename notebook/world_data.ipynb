## PHASE:2 Column Selection (Lazy)

Required Columns
```table
REF_AREA          → country_code      (dim_country)
REF_AREA_LABEL    → country_name      (dim_country)
INDICATOR         → indicator_code    (dim_indicator)
INDICATOR_LABEL   → indicator_name    (dim_indicator)
UNIT_MEASURE      → unit_measure      (dim_indicator)
TIME_PERIOD       → year              (dim_date)
SEX               → sex_code          (dim_sex)
SEX_LABEL         → sex_label         (dim_sex)
AGE               → age_code          (dim_age)
AGE_LABEL         → age_label         (dim_age)
URBANISATION      → urban_code        (dim_urbanisation)
URBANISATION_LABEL→ urban_label       (dim_urbanisation)
OBS_VALUE         → obs_value         (fact)
OBS_STATUS        → obs_status        (fact)
```

````python
# filtering the required columns
require_cols = [
    "REF_AREA",
    "REF_AREA_LABEL",
    "INDICATOR",
    "INDICATOR_LABEL",
    "UNIT_MEASURE",
    "TIME_PERIOD",
    "SEX",
    "SEX_LABEL",
    "AGE",
    "AGE_LABEL",
    "URBANISATION",
    "URBANISATION_LABEL",
    "OBS_VALUE",
    "OBS_STATUS",
]
#filtering all lazyframe
lf = lf.select(require_cols)
renaming_cols ={
    "REF_AREA": "country_code",     
"REF_AREA_LABEL" : "country_name",      
"INDICATOR": "indicator_code",
"INDICATOR_LABEL": "indicator_name",   
"UNIT_MEASURE" : "unit_measure",    
"TIME_PERIOD" : "year",            
"SEX" : "sex_code",      
"SEX_LABEL" : "sex_label",       
"AGE" : "age_code",          
"AGE_LABEL" : "age_label",       
"URBANISATION" : "urban_code",      
"URBANISATION_LABEL" : "urban_label",     
"OBS_VALUE" : "obs_value",         
"OBS_STATUS" : "obs_status"        
}
lf = lf.rename(renaming_cols)
print(f"new schema: {lf.collect_schema()}")
````

## PHASE:4 Extracting

````python
target_ind = [
    "WB_WDI_NY_GDP_PCAP_KD", "WB_WDI_NY_GDP_MKTP_KD_ZG",
    "WB_WDI_FP_CPI_TOTL_ZG", "WB_WDI_NE_TRD_GNFS_ZS",
    "WB_WDI_SP_DYN_LE00_IN", "WB_WDI_SH_DYN_MORT",
    "WB_WDI_SH_STA_MMRT", "WB_WDI_SH_XPD_CHEX_GD_ZS",
    "WB_WDI_SE_SEC_ENRR", "WB_WDI_SE_TER_ENRR",
    "WB_WDI_IT_NET_USER_ZS", "WB_WDI_IT_CEL_SETS_P2",
    "WB_WDI_SP_POP_TOTL", "WB_WDI_SP_POP_GROW",
    "WB_WDI_SL_UEM_TOTL_ZS", "WB_WDI_SL_TLF_CACT_ZS",
    "WB_WDI_EN_GHG_ALL_MT_CE_AR5", "WB_WDI_EG_USE_PCAP_KG_OE",
    "WB_WDI_EG_ELC_ACCS_ZS", "WB_WDI_EG_USE_COMM_FO_ZS"
]

wd_filter_ind = (lf.filter(pl.col("indicator_code").is_in(target_ind))
                    .filter(pl.col("year").cast(pl.Int32) <= 2023)
                    .filter(pl.col("obs_value").is_not_null())
                    .collect())
````

````python
print(wd_filter_ind.select([
    pl.col("country_code").null_count(),
    pl.col("indicator_code").null_count(),
    pl.col("year").null_count(),
    pl.col("sex_code").null_count(),
    pl.col("age_code").null_count(),
    pl.col("urban_code").null_count(),
]))
print(wd_filter_ind.select(["sex_code","age_code","urban_code"]).unique().sort("sex_code"))
````

````python
dim_country = wd_filter_ind.select(["country_code","country_name"]).unique().sort("country_code")
````
